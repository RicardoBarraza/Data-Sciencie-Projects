```python
# App para la predicción a través de SVM de ataque cardiaco para un paciente. 
```

```python
import pandas as pd
import numpy as np

import panel as pn
pn.extension()
```

# Crear los objetos de la interfaz

```python
edad = pn.widgets.IntSlider(name='Edad', start=29, end=77, step=1, value=55)
edad
```

```python
sexo = pn.widgets.RadioBoxGroup(name='Sexo', options=['Masculino', 'Femenino'], inline=True)
sexo
```

```python
cp = pn.widgets.RadioBoxGroup(name='Tipo de dolor en el pecho', options=['Típico de angina',
                                                                         'Atípico de angina',
                                                                         "No es dolor de angina",
                                                                        "Asintomático"], inline=True)
cp
```

```python
presion= pn.widgets.IntInput(name='Presion arterial', value=130, step=1, start=90, end=200)
presion
```

```python
colesterol = pn.widgets.IntSlider(name='Colesterol', start=126, end=564, step=1, value=240)
colesterol
```

```python
texto_diabético=pn.widgets.StaticText(value='El paciente es diabético:')
diabético = pn.widgets.RadioBoxGroup(name='Diabético', options=['Si', 'No'], inline=True)
diabético_caja =pn.WidgetBox(texto_diabético, diabético)
diabético_caja
```

```python
electro = pn.widgets.Select(name='Resultados del Electrocardiograma', 
                            options=['Normal', 'having ST-T', 
                                     'probable or definite left ventricular hypertrophy'])
electro
```

```python
frecuencia=pn.widgets.IntInput(name='Frecuencia cardiaca máxima', value=150, step=1, start=70, end=202)
frecuencia
```

```python
texto_angina=pn.widgets.StaticText(value='La angina fue por el ejercicio:')
ang_ejer = pn.widgets.RadioBoxGroup(name='Angina causada por ejercicio', options=['Si', 'No'], inline=True)
ang_ejer_caja =pn.WidgetBox(texto_angina, ang_ejer)
ang_ejer_caja 
```

```python
caa=pn.widgets.IntInput(name='Vasos principales corazón', value=3, step=1, start=0, end=4)
caa
```

```python
#Ventana del resultado / prediccion
resultado = pn.widgets.StaticText(name='Diagnóstio', value='esperando...')
resultado
```

```python
button = pn.widgets.Button(name='Predecir', button_type='primary')
button
```

```python
# Cargar el modelo y el mecanismo de escalamiento
import joblib

#1.- El modelo
modelo=joblib.load("modelo_ataque_cardiaco_svm.pkl")
#2.- Mecanismo de transformación de los datos
escala=joblib.load("escala_ataque_cardiaco_svm.pkl")
```

```python
# Definimos la función para hacer la prediccion.

def predecir(event):
    datos_nuevos=pd.DataFrame()
    datos_nuevos["age"]=[edad.value]
    datos_nuevos["sex"]=[int (np.where(sexo.value=="Femenino",1,0))]
    #Tipo de dolor en el pecho
    tipo_dolor={'Típico de angina':0,"Atípico de angina":1,"No es dolor de angina":3,"Asintomático":4}
    datos_nuevos["cp"]=[tipo_dolor[cp.value]]
    datos_nuevos["trtbps"]=[presion.value]
    datos_nuevos["chol"]=[colesterol.value]
    datos_nuevos["fbs"]=[int (np.where(diabético.value=="Si",1,0))]
    # Definir los valores del electrocardiagrama
    condiciones={"Normal":0,"having ST-T":1,"probable or definite left ventricular hypertrophy":2}
    datos_nuevos["restecg"]=[condiciones[electro.value]]
    datos_nuevos["thalachh"]=[frecuencia.value]
    datos_nuevos["exng"]=[int (np.where(ang_ejer.value=="Si",1,0))]
    datos_nuevos["caa"]=[caa.value]
    # Escalamos los datos
    datos_nuevos_trans=escala.transform(datos_nuevos)
    y_pred=int(modelo.predict(datos_nuevos_trans))
    if y_pred==1:
        resultado.value="Con altas probabilidades de tener un ataque cardicado"
    else:
        resultado.value="Con bajas probabilidades de tener un ataque cardicado"

        

button.on_click(predecir)
```

```python
objetos=pn.WidgetBox(edad,sexo,cp,presion,colesterol,diabético_caja,electro, frecuencia,ang_ejer_caja,caa,
                    button,resultado)
objetos
```

```python
# Interfaz
template=pn.template.FastListTemplate(
    title="Predicción de ataques cardiacos",
    sidebar=[pn.pane.Markdown("# Enfermedad cardiaca:"),
            pn.pane.Markdown("### Las enfermedades cardiacas son la segunda de fallecimientos.."),
            pn.pane.JPG("casas.jpeg")],  #<- Cambiar
    main=[objetos]
)
template.show()
```
