# Identificación de riesgo de una persona tenga un ataque cardiaco utilizando Naive Bayes

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, KFold, RepeatedKFold, cross_val_score
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

# Algortimo de Naive Bayes.
from sklearn.naive_bayes import MultinomialNB
```

```python
datos=pd.read_csv("heart_attack.csv")
```

```python
datos.head()
```

```python
datos[["age","trtbps","chol","thalachh"]].describe()
```

```python
# NOTA: Para aplicar Naive Bayes cuando en el dataframe tenemos atributos 
# discretos y continuos, la idea es transformar los atributos continuos
# a discretos creando clases de categorías.
```

```python
# Vamos a recodificar la edad
condiciones=[datos["age"]<40,(datos.age>=40)&(datos.age<60),datos["age"]>=60]
valores=[0,1,2]
datos["edad_c"]=np.select(condiciones,valores)
```

```python
# Vamos a recodificar la presion arterial
condiciones=[datos["trtbps"]<=130,datos["trtbps"]>130]
valores=[0,1]
datos["trtbps_c"]=np.select(condiciones,valores)
```

```python
# Vamos a recodificar los niveles de colesterol
condiciones=[datos["chol"]<=211,(datos["chol"]>211)&(datos["chol"]<274.5),datos["trtbps"]>=274.5]
valores=[0,1,2]
datos["chol_c"]=np.select(condiciones,valores)
```

```python
# Vamos a recodificar los niveles de frecuencia máxima cardiaca
condiciones=[datos["thalachh"]<=133.5,(datos["thalachh"]>133.5)&(datos["thalachh"]<153),
             (datos["thalachh"]>=153)&(datos["thalachh"]>=166),datos["thalachh"]>166]
valores=[0,1,2,3]
datos["thalachh_c"]=np.select(condiciones,valores)
```

```python
datos.head()
```

```python
# Seleccionamos el target
y=datos.output
# Seleccionamos los atributos
X=datos[["sex","cp","fbs","restecg","exng","caa","edad_c","trtbps_c","chol_c","thalachh_c"]]

```

```python
X.head()
```

```python
# Definimos el modelo
modelo=MultinomialNB()
scores=cross_val_score(modelo,X,y,scoring="recall",cv=10,n_jobs=-1)
scores
print("La sensibilidad promedio es {:.3f} +/- {:.3f}".format(scores.mean(), scores.std()))
```

```python
# Definimos el modelo
modelo=MultinomialNB()
scores=cross_val_score(modelo,X,y,scoring="accuracy",cv=10,n_jobs=-1)
scores
print("La sensibilidad promedio es {:.3f} +/- {:.3f}".format(scores.mean(), scores.std()))
```
